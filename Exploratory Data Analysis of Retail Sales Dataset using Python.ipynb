#Converting Total Sales and Profit to Indian Rupee Format

```python
from babel.numbers import format_currency

total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()

print("Total Sales:", format_currency(round(total_sales), 'INR', locale='en_IN'))
print("Total Profit:", format_currency(round(total_profit), 'INR', locale='en_IN'))
```
