```python
import pandas as pd
df=pd.read_excel(r"C:\Users\boomi\Downloads\superstore.xls")
df
```

# Sales by region (Bar Chart)

```python
import matplotlib.pyplot as plt

region_sales = df.groupby('Region')['Sales'].sum()

plt.figure()
region_sales.plot(kind='bar')
plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.show()
```

# Profit by Category (Pie chart)

```python
category_profit = df.groupby('Category')['Profit'].sum()

plt.figure()
plt.pie(category_profit, labels=category_profit.index, autopct='%1.1f%%')
plt.title('Profit by Category')
plt.show()
```

# Monthly Sales Trend (Line Chart)

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Month'] = df['Order Date'].dt.month

monthly_sales = df.groupby('Month')['Sales'].sum()

plt.figure()
plt.plot(monthly_sales)
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.show()
```

# Discount vs Profit (Scatter Plot)

```python
plt.figure()
plt.scatter(df['Discount'], df['Profit'])
plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.show()
```

# Sales Distribution (Histogram)

```python
plt.figure()
plt.hist(df['Sales'], bins=20)
plt.title('Sales Distribution')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()
```

## Visualization Insights


1. The bar chart of sales by region shows that the West region contributes the highest share of total sales among all regions.

2. Visualization of profit by category highlights that the Technology category generates the highest profit compared to Furniture and Office Supplies.

3. The monthly sales line chart reveals a clear seasonal pattern, with sales peaking towards the end of the year, particularly in November.

4. The scatter plot between discount and profit indicates a negative relationship, where higher discount values are generally associated with lower profits.

5. The sales distribution histogram shows that most orders fall into the lower sales range, with only a few high-value orders contributing significantly to total sales.
